```python
# DATA BOOTSTRAP — run once per session after setup cell
# WHY: Colab's filesystem resets each session. This downloads the NYC TLC
# Yellow Taxi dataset — a real-world financial dataset with 19 columns
# including fare_amount, tip_amount, payment_type, and datetime fields.
# It replaces the synthetic sample files from the original repo which
# were too small and too simple to demonstrate meaningful PySpark operations.
# Run this notebook first. All subsequent notebooks reference /content/data/.
import os, shutil, subprocess

os.makedirs('/content/data', exist_ok=True)

# Download NYC TLC Yellow Taxi Jan 2023 (~50MB, ~3M rows)
if not os.path.exists('/content/data/trips.parquet'):
    os.system('wget -q https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2023-01.parquet -O /content/data/trips.parquet')

trips = spark.read.parquet('/content/data/trips.parquet')

# CSV version for CSV exercises
trips.limit(10000).write.mode('overwrite').option('header', True).csv('/content/data/trips_csv')

# JSON version for JSON exercises
trips.limit(1000).write.mode('overwrite').json('/content/data/trips_json')

# Text version for word count / text exercises
with open('/content/data/data.txt', 'w') as f:
    f.write('hello world\nhello spark\nspark is fast\nworld of spark\nhello hello world\n')

print(f'Dataset ready: {trips.count():,} rows, {len(trips.columns)} columns')
trips.printSchema()
```
